# 🧙‍♂️ Day 2: Spell Casting Loops at Hogwarts

```python
spells = ["Expelliarmus", "Stupefy", "Lumos", "Accio", "Protego"]

for spell in spells:
    print(f"Casting {spell}! ✨")
```

```python
mana = 5

while mana > 0:
    print(f"You have {mana} mana left. Casting spell... 💥")
    mana -= 1

print("You're out of mana! Drink a potion 🧪")
```

```python
import random

opponent = random.choice(["Draco", "Voldemort", "Bellatrix"])
print(f"You are dueling with {opponent}! 🧙‍♀️⚔️")

spells = ["Expelliarmus", "Stupefy", "Protego", "Sectumsempra"]

for round_num in range(1, 4):
    print(f"\nRound {round_num} - Choose your spell:")
    print(spells)
    
    user_spell = input("Type your spell: ")

    if user_spell in spells:
        print(f"💥 You cast {user_spell} at {opponent}!")
    else:
        print("🧹 You fumbled your wand! No spell cast.")
```

# 🔮 Bonus Challenge:
- Add HP (Health Points) for both player and opponent
- Reduce HP on spell hit
- End with a win/lose message 🏆


# 🧪 Hogwarts Duel With HP – Final Battle Code

```python
import random

# Setup duel
opponent = random.choice(["Draco", "Voldemort", "Bellatrix"])
print(f"You are dueling with {opponent}! 🧙‍♂️⚔️")

# Initial HP
your_hp = 100
opponent_hp = 100

# Spell list with possible damage values
spells = {
    "Expelliarmus": (10, 20),
    "Stupefy": (15, 25),
    "Protego": (0, 0),  # Defense spell
    "Sectumsempra": (20, 35)
}

# Duel begins
round_num = 1

while your_hp > 0 and opponent_hp > 0:
    print(f"\n--- Round {round_num} ---")
    print(f"Your HP: {your_hp} | {opponent}'s HP: {opponent_hp}")
    print("Available spells:", list(spells.keys()))
    
    spell = input("Choose your spell: ").strip()

    if spell in spells:
        dmg_range = spells[spell]
        if dmg_range == (0, 0):
            print("🛡️ You defended this round with Protego!")
        else:
            damage = random.randint(*dmg_range)
            opponent_hp -= damage
            print(f"💥 You hit {opponent} with {spell}, causing {damage} damage!")
    else:
        print("❌ Invalid spell. You lost your turn!")

    # Opponent attacks
    if opponent_hp > 0:
        opp_damage = random.randint(10, 25)
        your_hp -= opp_damage
        print(f"🔥 {opponent} strikes back! You lose {opp_damage} HP.")

    round_num += 1

# End Game
print("\n🏁 Duel Over!")
if your_hp > 0 and opponent_hp <= 0:
    print(f"🏆 You defeated {opponent} and won the duel!")
elif opponent_hp > 0 and your_hp <= 0:
    print(f"💀 You were defeated by {opponent}. Better luck next time!")
else:
    print("⚖️ It's a draw! Both duelers fell at the same time.")
```
