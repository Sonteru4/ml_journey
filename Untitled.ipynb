```python
# Voice Input
import speech_recognition as sr

def get_audio():
    r = sr.Recognizer()
    with sr.Microphone() as source:
        print("Listening...")
        audio = r.listen(source)
        try:
            return r.recognize_google(audio)
        except:
            return "Sorry, I didn't catch that."
```

```python
# Voice Output
import pyttsx3

engine = pyttsx3.init()
def speak(text):
    engine.say(text)
    engine.runAndWait()
```

```python
import openai

openai.api_key = "your-api-key"

def chat_with_gpt(prompt):
    response = openai.ChatCompletion.create(
        model="gpt-4",
        messages=[{"role": "user", "content": prompt}]
    )
    return response['choices'][0]['message']['content']
```

```python
from datetime import datetime
import requests

def get_weather():
    city = "New York"
    api_key = "your-weather-api-key"
    res = requests.get(f"http://api.weatherapi.com/v1/current.json?key={api_key}&q={city}")
    return f"The current temperature is {res.json()['current']['temp_f']} °F in {city}."

def daily_digest():
    now = datetime.now().strftime("%A, %B %d")
    weather = get_weather()
    speak(f"Good morning! Today is {now}. {weather}")
```

```python
import schedule
import time

schedule.every().day.at("08:00").do(daily_digest)

while True:
    schedule.run_pending()
    time.sleep(1)
```
